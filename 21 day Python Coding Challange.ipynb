#### 3) Write a Python program to count even and odd numbers in a list using a function.

```python
def count_even_odd(numbers):
    even = 0
    odd = 0

    for num in numbers:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1

    return even, odd
values = [1,2,3,4,5,6]
even_count, odd_count = count_even_odd(values)

print("Even numbers: ", even_count)
print("Odd numbers: ", odd_count)
```

#### 1) Build a menu-driven Python program to manage student records.Each student should have an ID, name, and marks.
#### 


```python
students = []

def add_student():
    try:
        sapid = int(input("Enter student ID: "))
        name = input("Enter student name: ")
        marks = float(input("Enter marks: "))

        student = {
            "id": sapid,
            "name": name,
            "marks": marks
        }
        students.append(student)
        print("Student added successfully.")
    except ValueError:
        print("Invalid input. Please enter correct values.")

def view_students():
    if not students:
        print("No student records found.")
        return

    for s in students:
        print("ID:", s["id"], "| Name:", s["name"], "| Marks:", s["marks"])

def search_student():
    sid = int(input("Enter student ID to search: "))
    for s in students:
        if s["id"] == sid:
            print("Student found:", s)
            return
    print("Student not found.")

while True:
    print("\nMenu")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        view_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        print("Exiting program.")
        break
    else:
        print("Invalid choice.")
```
