### 1. Add few names, one name in each row, in “name.txt file”.
#### a. Count no of names
#### b. Count all names starting with vowel
### c. Find longest name

```python
with open("name.txt", "w") as f:
    f.write("Hitesh\n")
    f.write("Viraj\n")
    f.write("Devanshi\n")
    f.write("Om\n")
    f.write("Uday\n")
```

a)

```python
with open("name.txt", "r") as f:
    names = f.read().splitlines()

print("Total names:", len(names))
```

b)

```python
with open("name.txt", "r") as f:
    names = f.read().splitlines()

vowels = "AEIOUaeiou"
count = 0

for name in names:
    if name[0] in vowels:
        count += 1

print("Names starting with vowel:", count)
```

c)

```python
with open("name.txt", "r") as f:
    names = f.read().splitlines()

longest = max(names, key=len)
print("Longest name:", longest)
```

### 2. Store integers in a file.
### a. Find the max number
### b. Find average of all numbers
### c. Count number of numbers greater than 100


```python
with open("numb.txt", "w") as f:
    f.write("20\n100\n200\n35\n550\n")
```

a

```python
with open("numb.txt", "r") as f:
    nums = list(map(int, f.read().split()))

print("Maximum:", max(nums))
```

b

```python
with open("numb.txt", "r") as f:
    nums = list(map(int, f.read().split()))

avg = sum(nums) / len(nums)
print("Average:", avg)
```

c

```python
with open("numb.txt", "r") as f:
    nums = list(map(int, f.read().split()))

count = 0
for n in nums:
    if n > 100:
        count += 1

print("Numbers > 100:", count)
```

### 4. Input two values from user where the first line contains N, the number of test
### cases. The next N lines contain the space separated values of a and b. Perform
### integer division and print a/b. Handle exception in case of ZeroDivisionError or
### ValueError.
### Sample input
### 1 0
### 2 $
### 3 1
### Sample Output:
### Error Code: integer division or modulo by zero
### Error Code: invalid literal for int() with base 10: '$' 3

```python
n = int(input("Enter number of test cases: "))

for i in range(n):
    try:
        a, b = input().split()
        print(int(a) // int(b))
    except ZeroDivisionError as e:
        print("Error Code:", e)
    except ValueError as e:
        print("Error Code:", e)
```

### 5. Create multiple suitable exceptions for a file handling program.


```python
with open("data.txt", "w") as f:
    f.write("10 20 30")
```

```python
class FileError(Exception): pass

try:
    data = open("data.txt").read().strip()

    if not data:
        raise FileError("Empty File")

    if not data.replace(" ","").isdigit():
        raise FileError("Invalid Data")

    print("Valid Data")

except Exception as e:
    print("Error:", e)
```
