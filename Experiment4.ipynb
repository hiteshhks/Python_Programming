### Q1) Find a factorial of given number.

```python
number = int(input("Enter a number: "))

factorial = 1
for i in range(1, number + 1):
    factorial *= i

print("Factorial of", number, "is =", factorial)
```

### Q2) Find whether the given number is Armstrong number.

```python
num = int(input("Enter a number: "))

original_number = num

sum_of_digits = 0

while num > 0:
    digit = num % 10
    sum_of_digits += digit ** 3
    num //= 10

if sum_of_digits == original_number:
    print(original_number, "It is an Armstrong number")
else:
    print(original_number, "It is not an Armstrong number")
```

### Q3) Print Fibonacci series up to given term


```python
terms = int(input("Enter number of terms: "))
first = 0
second = 1

print("Fibonacci Series of:",terms)

for i in range(terms):
    print(first, end=" ")
    next_term = first + second
    first = second
    second = next_term
```

### Q4) Write a program to find if given number is prime number or not.


```python
number = int(input("Enter a number: "))

is_prime = True

if number <= 1:
    is_prime = False
else:
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

if is_prime:
    print(number, "is a prime number")
else:
    print(number, "is not a prime number")
```

### Q5) Check whether given number is palindrome or not.

```python
number = int(input("Enter a number: "))

original_number = number
reversed_number = 0

while number > 0:
    digit = number % 10
    reversed_number = reversed_number * 10 + digit
    number //= 10

if original_number == reversed_number:
    print(original_number, "is a palindrome number")
else:
    print( original_number, "is not a palindrome number")
```

### Q6) Write a program to print sum of digits


```python
number = int(input("Enter a number: "))

sum_of_digits = 0

while number > 0:
    digit = number % 10
    sum_of_digits += digit
    number //= 10

print("Entered number =", sum_of_digits)
print("Sum of digits =", sum_of_digits)
```

### Q7) Count and print all numbers divisible by 5 or 7 between 1 to 100.

```python
count = 0

print("Numbers divisible by 5 or 7 between 1 and 100:")

for i in range(1, 101):
    if i % 5 == 0 or i % 7 == 0:
        print(i, end=" ")
        count += 1

print("\nTotal count =", count)
```

### Q9) Print all prime numbers between 1 and 100.

```python
print("Prime numbers between 1 and 100:")

for num in range(2, 101):
    is_prime = True
    
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print(num, end=" ")
```
